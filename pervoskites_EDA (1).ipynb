```python
import numpy as np
import plotly.express as px

# Select only numeric columns
numeric_data = data.select_dtypes(include=[np.number])

# Compute correlation matrix
corr_matrix = numeric_data.corr()

# Plot heatmap with a larger figure size
fig = px.imshow(corr_matrix, text_auto=True, color_continuous_scale="RdBu")

# Increase figure size
fig.update_layout(title="Feature Correlation Heatmap", width=1200, height=1000)

fig.show()
```

```python
fig = px.box(data, y=["PBE band gap", "μ", "μĀ", "μ𝐵 ̅"], title="Boxplot of Key Features")
fig.show()
```

```python
fig = px.histogram(data, x='stable', nbins=3, title="Distribution of Stability")
fig.show()
```

```python
fig = px.bar(data, x="functional group", title="Count of Functional Groups",
             width=1000, height=600)
fig.show()
```

```python
fig = px.scatter(data, x="PBE band gap", y="stable", color="stable",
                 title="PBE Band Gap vs Stability")
fig.show()
```

```python
fig = px.scatter_3d(data, x="PBE band gap", y="μ", z="stable", color="Insulator",
                    title="3D Scatter Plot of Band Gap, μ, and Stability")
fig.show()
```

```python
fig = px.histogram(data, x="PBE band gap", nbins=50,
                   title="Distribution of PBE Band Gap",
                   color_discrete_sequence=['blue'],
                   width=900, height=500)
fig.show()
```

```python
fig = px.box(data, y="PBE band gap", title="Boxplot of PBE Band Gap",
             width=800, height=500)
fig.show()
```

```python
import numpy as np

# Select numerical columns
numeric_data = data.select_dtypes(include=[np.number])

# Compute correlation
corr_matrix = numeric_data.corr()

# Get top correlated features with Band Gap
corr_bandgap = corr_matrix["PBE band gap"].sort_values(ascending=False)

# Plot bar chart
fig = px.bar(x=corr_bandgap.index, y=corr_bandgap.values,
             title="Correlation of Features with PBE Band Gap",
             labels={'x': 'Features', 'y': 'Correlation'},
             width=1000, height=500)
fig.show()
```

```python
fig = px.scatter_3d(data, x="PBE band gap", y="stable", z="Insulator",
                    color="stable",
                    title="3D Scatter Plot of Band Gap, Stability, and Insulator",
                    width=1000, height=800)
fig.show()
```
